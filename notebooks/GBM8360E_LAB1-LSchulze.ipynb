### D)
Both formats store image geometry and physics parameters that are necessary for image reconstruction and analysis (e.g. matrix dimensions, voxel sizes, TR, TE, flip angle). <br>
The DICOM Header is more clinically focused and contains administrative data, patient demographics and scanner hardware details. Whereas the NIfTI Header is optimized for data processing. The file header focuses on 3D spacial orientation, while the JSON sidecar retains only the relevant physics parameters for research and stripping away the clinical metadata. <br>

### E)
Based on the acquisition parameters, the images are classified as either T1-weighted or T2-weighted due to their specific TR and TE values. <br>
A T1-weighted acquisition uses a short TR and a short TE. The short TR maximizes T1 contrast by capturing the signal before full magnitisation recovery, while the short TE minimizes T2 effects. This specific combination results in an image contrast where fluids appear dark and fat or white matter appears bright. <br>
A T2-weighted acquisition utilizes a long TR and a long TE. The long TR minimizes T1 effects by allowing all tissues to fully recover, whereas the long TE maximizes T2 contrast by highlighting differences in signal decay. This procedures an image contrast where fluids appear very bright and white matter appears darker than gray matter.

### C)
**Property A (Sampling):**
Appears when the Field of View (FOV) is set smaller than the physical anatomy being scanned. It is not recoverable from a single-coil image. The tissue signals are physically superimposed, making true recovery impossible without multi-coil parallel imaging data. <br>

**Property B (Rectangular window):**
Appears in almost all scans because MRI systems must physically stop sampling at some point, resulting in a finite k-space matrix (a rectangular cutoff). <br>

**Property C (Hermitian Symmetry):**
Occurs when the scanner acquires only sligthly more than half of k-space to reduce scan time. It is recoverable with algorithms that mathematically synthesize the missing half using the acquired phase data. Though, it is not bitwise identical, as reconstructing the data mathematically changes the noise profile, resulting in a slightly lower Signal-to-Noise Ratio (SNR) than a fully acquired scan. <br>

**Property H (Addition property):**
Happens due to random static electrical discharges in the scanner room or unshielded external RF interference (e.g. open scanner door), registering as a massive artificial impulse in k-space. It is recoverable. You can locate the extreme spike in the raw k-space matrix and replace it with the mathematical average of its immediate neighbours. It is not bitwise identical since the true anatomical data at that exact k-space coordinate was overwritten by the spark and is replaced by an estimation. <br>

**Property J (k-space shift):**
Appears due to physical hardware imperfections like eddy currents or gradient delays, which slightly shift the centre of the k-space acquisition. It is recoverable by shifting the k-space matrix back to its true centre. Also it is bitwise identical since spatical shifting is a completly reversible linear mathematical operation, meaning no original data is lost or altered.

### D)⁠
I chose to demonstrate the Translation property, which is based on the Fourier Shift Theorem. First, the magnitude and phase images were combined into a single complex data array using Euler's formula to accurately represent the MR signal. Next, a 2D Fast Fourier Transform was applied to convert this complex image from the spatial domain into the frequency domain (k-space).According to the shift theorem, translating an object in the spatial domain corresponds to multiplying its k-space representation by a linear phase ramp. Therefore, a spatial shift was induced by generating a 2D phase ramp matrix and multiplying it with the k-space data. Finally, an inverse FFT was performed to transform the modified data back into the spatial domain, where the absolute value and the angle were extracted to generate the new, spatially shifted magnitude and phase images.

```python
import numpy as np
import nibabel as nib
import matplotlib.pyplot as plt

# 1. Load magnitude and phase data
mag_data = nib.load('magnitude.nii').get_fdata()
phase_data = nib.load('phase.nii').get_fdata()

# Select one 2D slice
mag_slice = mag_data[:, :, mag_data.shape[2]//2]
phase_slice = phase_data[:, :, phase_data.shape[2]//2]

# 2. Combine to complex image and transform to k-space
complex_img = mag_slice * np.exp(1j * phase_slice)
k_space = np.fft.fftshift(np.fft.fft2(complex_img))

# 3. Apply Translation Property (Fourier Shift Theorem)
nx, ny = k_space.shape
x = np.arange(-nx//2, nx//2)
y = np.arange(-ny//2, ny//2)
X, Y = np.meshgrid(y, x)

# Define shift 
shift_x, shift_y = 30, 40 
phase_ramp = np.exp(-1j * 2 * np.pi * ((X * shift_x / nx) + (Y * shift_y / ny)))

# Multiply k-space by phase ramp
k_space_shifted = k_space * phase_ramp

# 4. Transform back to image space
img_shifted = np.fft.ifft2(np.fft.ifftshift(k_space_shifted))
new_mag = np.abs(img_shifted)
new_phase = np.angle(img_shifted)

# 5. Plot results
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(new_mag, cmap='gray')
ax[0].set_title('Manipulated Magnitude (Shifted)')
ax[1].imshow(new_phase, cmap='gray')
ax[1].set_title('Manipulated Phase (Shifted)')
plt.show()
```

## EXERCISE 3
### A)⁠
I loaded the raw complex k-space array and calculated its magnitude. Because in k-space data the centre contains the majority of the signal while the edges are very faint, applying a logarithmic transformation compresses the intensity scale. This makes the high-frequency information in the periphery visible to the human eye. <br>
I used a 2D Inverse Fast Fourier Transform to convert the spatial frequencies back into an anatomical image. I applied 'ifftshift' before the transformation to move the low frequencies from the centre to the corners and 'fftshift' afterward to correctly re-centre the reconstructed image. <br>
From the resulting complex image array I calculated the absolute value to generate the standard magnitude image (anatomy) and extracted the complex angle to display the phase image.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Load the data
file_path = 'kspace_combined.npy'
k_space = np.load(file_path) 

# 2. Calculate Magnitude and Log Magnitude of k-space
k_space_mag = np.abs(k_space)
k_space_log_mag = np.log(k_space_mag + 1e-5) # + constant to avoid log(0)

# 3. Transform to Image Space
# Shift center to corners -> Inverse FFT -> Shift corners back to center
img_space = np.fft.fftshift(np.fft.ifft2(np.fft.ifftshift(k_space)))

# 4. Extract Magnitude and Phase of the reconstructed image
img_mag = np.abs(img_space)
img_phase = np.angle(img_space)

# 5. Visualize the results
fig, ax = plt.subplots(2, 2, figsize=(10, 8))
ax[0, 0].imshow(k_space_mag, cmap='gray')
ax[0, 0].set_title('Magnitude of k-space')

ax[0, 1].imshow(k_space_log_mag, cmap='gray')
ax[0, 1].set_title('Log Magnitude of k-space')

ax[1, 0].imshow(img_mag, cmap='gray')
ax[1, 0].set_title('Image Space: Magnitude')

ax[1, 1].imshow(img_phase, cmap='gray')
ax[1, 1].set_title('Image Space: Phase')

plt.tight_layout()
plt.show()
```

### B)
The overall brightness of the image drastically drops and severe pixelation and ripple patterns appear and obscure the fine edges. <br>
The centre of k-space holds the low spatial frequencies, which provide the bulk of the image's signal power, overall brightness and basic contrast. Removing it causes the severe loss of brightness. Applying a harsh, rectangular mask creates a sudden cutoff in k-space. This mathematical discontinuity translates into severe truncation artifacts in the image domain, which visually present as pixelation and obscure the remaining anatomical details. <br>
The maximum extend of k-space is technically maintained, meaning the potential for high resolution remains. However, the artificial ringing from the hard mask dominates the image, making fine details visually indiscernible. <br>
The distance between samples points is completely unaffected by zeroing out values. Therefore, the spatial dimensions (FOV) of the reconstructed image remain identical to the original.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Load the k-space data
k_space = np.load('kspace_combined.npy')

# 2. Create the mask (zeroing out the central 1/2 of both dimensions)
rows, cols = k_space.shape
mask = np.ones((rows, cols))

r_start, r_end = rows // 4, 3 * rows // 4
c_start, c_end = cols // 4, 3 * cols // 4
mask[r_start:r_end, c_start:c_end] = 0

# 3. Apply the mask to the center of k-space
k_space_masked = k_space * mask

# 4. Transform back to image space
img_masked = np.fft.fftshift(np.fft.ifft2(np.fft.ifftshift(k_space_masked)))
img_mag = np.abs(img_masked)

# 5. Visualize 
fig, ax = plt.subplots(1, 2, figsize=(10, 5))

ax[0].imshow(img_mag, cmap='gray')
ax[0].set_title('Image Magnitude')

ax[1].imshow(np.log(np.abs(k_space_masked) + 1e-5), cmap='gray')
ax[1].set_title('Masked k-space')

plt.show()
```

### C)
The resulting image is halved in size along the downsampled dimension and the anatomy appears to overlap itself. This overlapping is known as aliasing. <br>
By sampling only every second row in k-space the distance between sampled points ($\Delta$k) is doubled. According to the equation doubling $\Delta$k halves the FOV. Because the object is now larger than this reduced FOV, the signal lying outside the boundary wraps around to the opposite side of the image. <br>
Even though there were lines skipped, it is still sampled all the way out to the original outer edges of k-space. Therefore, the maximum k-space extend remains unchanged. This means the intrinsic spatial resolution is preserved, even though the image is corrupted by aliasing.

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Load the k-space data
k_space = np.load('kspace_combined.npy')

# 2. Downsample k-space by half in one direction (take every 2nd row)
k_space_downsampled = k_space[::2, :]

# 3. Transform back to image space
img_downsampled = np.fft.fftshift(np.fft.ifft2(np.fft.ifftshift(k_space_downsampled)))
img_mag = np.abs(img_downsampled)

# 4. Visualize
fig, ax = plt.subplots(1, 2, figsize=(10, 5))

ax[0].imshow(img_mag, cmap='gray')
ax[0].set_title('Image Magnitude')

ax[1].imshow(np.log(np.abs(k_space_downsampled) + 1e-5), cmap='gray')
ax[1].set_title('Downsampled k-space')

plt.show()
```

### D)
I transformed the original k-space into spatial image domain and then simulated a 20-degree head tilt by rotating the image. Next, converted the rotated image back into k-space and combined the arrays. Lines before the simulated motion (top to centre) use the original data, lines after (centre to bottom) use the rotated data. I applied an Inverse FFT to the composite k-space to reveal the motion corrupted image. <br>
The resulting image exhibits severe blurring, overlapping anatomy and heavy wave-like distortions. Splicing creates a massive mathematical jump exactly at the red line in k-space. Because the motion accured near the k-space centre, the mismatch catastrophically destroys the overall coherence of the image. Sudden jumps in k-space translate to intense ringing artifacts spread across the phase-encoding direction.

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# 1. Load k-space and reconstruct original image
k_space_orig = np.load('kspace_combined.npy')
img_orig = np.fft.fftshift(np.fft.ifft2(np.fft.ifftshift(k_space_orig)))

# 2. Function to rotate matrix using PIL
def rotate_matrix(mat, angle):
    pil_img = Image.fromarray(mat.astype(np.float32))
    return np.array(pil_img.rotate(angle), dtype=np.float64)

# Apply 20-degree rotation (real and imaginary separated)
img_rot_real = rotate_matrix(np.real(img_orig), 20)
img_rot_imag = rotate_matrix(np.imag(img_orig), 20)
img_rot = img_rot_real + 1j * img_rot_imag

# 3. Transform rotated image back to k-space
k_space_rot = np.fft.fftshift(np.fft.fft2(np.fft.ifftshift(img_rot)))

# 4. Splice k-spaces (Simulate motion 5 lines after the center)
rows, cols = k_space_orig.shape
split_line = (rows // 2) + 5  

k_space_composite = np.zeros_like(k_space_orig, dtype=complex)
k_space_composite[:split_line, :] = k_space_orig[:split_line, :]
k_space_composite[split_line:, :] = k_space_rot[split_line:, :]

# 5. Reconstruct final corrupted image
img_simulated = np.fft.fftshift(np.fft.ifft2(np.fft.ifftshift(k_space_composite)))
img_mag_simulated = np.abs(img_simulated)

# 6. Visualize
fig, ax = plt.subplots(1, 2, figsize=(10, 5))

ax[0].imshow(np.log(np.abs(k_space_composite) + 1e-5), cmap='gray')
ax[0].set_title('Composite k-space (Sudden Shift)')
ax[0].axhline(y=split_line, color='r', linestyle='--', label='Motion occurs')
ax[0].legend()

ax[1].imshow(img_mag_simulated, cmap='gray')
ax[1].set_title('Image Magnitude (Motion Artifact)')

plt.show()
```
